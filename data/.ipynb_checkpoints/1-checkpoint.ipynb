```python
import pandas as pd
import os
import re

df = pd.read_csv("posts.csv")
images = os.listdir("raw_images")
```

```python
def extract_shortcode(filename):
    match = re.match(r"([A-Za-z0-9_-]+)", filename)
    return match.group(1) if match else None
```

```python
rows = []

for filename in images:

    if filename.endswith(".jpg"):

        shortcode = extract_shortcode(filename)

        if shortcode is None:
            continue

        match = df[df["shortcode"] == shortcode]

        if not match.empty:

            post = match.iloc[0]

            rows.append({
                "image_path": os.path.join("images", filename),
                "shortcode": shortcode,
                "likes": post["like_count"],
                "timestamp": post["taken_at"],
                "caption": post["caption"],
                "comment_count": post["comment_count"]
            })

final_df = pd.DataFrame(rows)
```

```python
import pandas as pd

final_df=pd.read_csv('labeled_dataset.csv')
final_df["category"] = None
```

```python
categories = [
    "match_action",
    "celebration",
    "player_portrait",
    "training",
    "fans_stadium",
    "graphic_post",
    "behind_scenes",
    "commercial"
]
```

```python
for idx, row in final_df.iterrows():

    if pd.isna(final_df.loc[idx, "category"]):

        print("\n")
        print("IMAGE:", row["image_path"])
        print("Likes:", row["likes"])

        print("\nCategories:")
        for i, c in enumerate(categories):
            print(f"{i}: {c}")

        choice = input("Select category number: ")

        if choice.isdigit():

            final_df.loc[idx, "category"] = categories[int(choice)]

            # 🔥 guardar automáticamente
            final_df.to_csv("labeled_dataset.csv", index=False)

            print("Saved!")
```

```python
import pandas as pd

final_df = pd.read_csv("labeled_dataset.csv")
final_df["category"].value_counts()
```

```python
if "category" not in final_df.columns:
    final_df["category"] = None
```

```python
final_df[["image_path", "category"]].head(20)
```

```python
categories = [
    "match_action",
    "celebration",
    "player_portrait",
    "training",
    "fans_stadium",
    "graphic_post",
    "behind_scenes",
    "commercial"
]
```

```python
##### from PIL import Image
import matplotlib.pyplot as plt
import os
import pandas as pd

for idx, row in final_df.iterrows():

    if pd.isna(row["category"]):

        image_path = os.path.join(
            "raw_images",
            row["image_path"]
        )

        # 🔥 comprobar si existe
        if not os.path.exists(image_path):

            continue

        try:

            img = Image.open(image_path)

            plt.imshow(img)
            plt.axis("off")
            plt.show()

            print("Likes:", row["likes"])

            print("\nCategories:")

            for i, c in enumerate(categories):
                print(f"{i}: {c}")

            choice = input("Select category number: ")

            if choice.isdigit():

                final_df.loc[idx, "category"] = categories[int(choice)]

                final_df.to_csv(
                    "labeled_dataset.csv",
                    index=False
                )

                print("Saved!")

        except Exception as e:

            print(f"Error loading image: {e}")
```
