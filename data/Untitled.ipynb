```python
import pandas as pd 
df=pd.read_csv('multilabel_dataset.csv')
```

```python
labels = [
    "PLAYER",
    "ACTION",
    "GRAPHIC",
    "CROWD",
    "CELEBRATION"
]
```

```python
from PIL import Image
import matplotlib.pyplot as plt

for idx, row in df.iterrows():

    if row[labels].sum() == 0:

        print("\nIMAGE:", row["image_path"])
        print("Likes:", row["likes"])

        img = Image.open(row["image_path"])
        plt.imshow(img)
        plt.axis("off")
        plt.show()

        print("\nLabels:")
        for i, label in enumerate(labels):
            print(f"{i}: {label}")

        choices = input("Select labels: ")

        if choices.strip() != "":

            for c in choices.split(","):

                c = c.strip()

                if c.isdigit():

                    i = int(c)

                    if i < len(labels):

                        df.loc[idx, labels[i]] = 1

        df.to_csv("multilabel_dataset.csv", index=False)
        print("Saved ✔")
```
